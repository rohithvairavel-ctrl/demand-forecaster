# 01 — EDA: Daily Demand

Explore the skforecast simulated item sales series: trends, weekly seasonality, and stationarity cues.

```python
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path('..').resolve()
sys.path.insert(0, str(ROOT / 'src'))
from demand_forecaster.data import load_demand

df, y = load_demand(series='item_1')
df.head(), y.describe()
```

```python
fig, axes = plt.subplots(2, 1, figsize=(12, 6), sharex=False)
y.plot(ax=axes[0], title='Daily demand (item_1)')
y.groupby(y.index.dayofweek).mean().plot(kind='bar', ax=axes[1], title='Avg by day-of-week')
plt.tight_layout()
plt.show()
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose
decomp = seasonal_decompose(y.asfreq('D').interpolate(), model='additive', period=7)
decomp.plot()
plt.tight_layout()
plt.show()
```
